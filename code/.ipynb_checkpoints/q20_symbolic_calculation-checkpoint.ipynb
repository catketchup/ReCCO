```python
class BubbleCollision_RF_symbolic():
    def __init__(self, A, B, r_H, chi_c, chi_e, Delta*chi_dec):
        self.A = A
        self.B = B
        self.r_H = r_H
        self.chi_c = chi_c
        self.chi_e = chi_e
        self.Delta*chi_dec = Delta*chi_dec
        
    def Cos_theta_c(self):
        str_Cos_theta_c = '(chi_c - chi_e*cos(theta_e))/Delta*chi_dec'
        return sympify(str_Cos_theta_c, evaluate=False)

    def Delta_cos_theta_c_n(self, n):
        str_Delta_cos_theta_c_n = f'1 - ({Cos_theta_c()})**{n}'
        return sympify(str_Delta_cos_theta_c_n, evaluate=False)

    # def Cos_theta_c(self, theta_e):
    #     return (self.chi_c - self.chi_e*cos(theta_e))/self.Delta*chi_dec
        
    # def Delta_cos_theta_c_n(self, cos_theta_c, n):
    #     return 1 - cos_theta_c**n

    def RQF_SW(self):
        cos_theta_c = self.Cos_theta_c()
        Delta_cos_theta_c_n = self.Delta_cos_theta_c_n()
        
        sin_theta_e = sin(theta_e)
        cos_theta_e = cos(theta_e)

        str_RQF_SW = f'(2*D_psi_dec - 3/2)*5*sqrt(6)/16*sin_theta_e**2*\
        (A/r_H*(3/4*Delta*chi_dec*({1 - cos_theta_c**4})\
                     + (chi_e*cos_theta_e-chi_c)*({Delta_cos_theta_c_n(3)}) \
                     -1/2*Delta*chi_dec*({Delta_cos_theta_c_n(2)}) - \
                     (chi_e*cos_theta_e - chi_c)* (1 - cos_theta_c)) + \
         B/(r_H**2)*(3/5*Delta*chi_dec**2*({Delta_cos_theta_c_n(5)}) + \
                          3/2*Delta*chi_dec*(chi_e*cos_theta_e-chi_c)*({Delta_cos_theta_c_n(4)}) + \
                          1/3*(-Delta*chi_dec**2 + \
                               3*(chi_e*cos_theta_e-chi_c)**2)*({Delta_cos_theta_c_n(3)}) - \
                          Delta*chi_dec*(chi_e*cos_theta_e-chi_c)*({Delta_cos_theta_c_n(2)}) - \
                          (chi_e*cos_theta_e-chi_c)**2*({Delta_cos_theta_c_n(1)})))'

        return sympify(str_RQF_SW, evaluate=False)
        
        # return (Symbol('2')*D_psi_dec-Symbol('3/2'))*5*np.sqrt(6)/16*sin_theta_e**2*\
        # (A/self.r_H*(3/4*self.Delta*chi_dec*self.Delta_cos_theta_c_n(cos_theta_c, 4)\
        #              + (self.chi_e*cos_theta_e-self.chi_c)*self.Delta_cos_theta_c_n(cos_theta_c, 3) \
        #              -1/2*self.Delta*chi_dec*self.Delta_cos_theta_c_n(cos_theta_c, 2) - \
        #              (self.chi_e*cos_theta_e -self.chi_c)* self.Delta_cos_theta_c_n(cos_theta_c, 1)) + \
        #  B/(self.r_H**2)*(3/5*self.Delta*chi_dec**2*self.Delta_cos_theta_c_n(cos_theta_c, 5) + \
        #                   3/2*self.Delta*chi_dec*(self.chi_e*cos_theta_e-self.chi_c)*self.Delta_cos_theta_c_n(cos_theta_c, 4) + \
        #                   1/3*(-self.Delta*chi_dec**2 + \
        #                        3*(self.chi_e*cos_theta_e-self.chi_c)**2)*self.Delta_cos_theta_c_n(cos_theta_c,3) - \
        #                   self.Delta*chi_dec*(self.chi_e*cos_theta_e-self.chi_c)*self.Delta_cos_theta_c_n(cos_theta_c, 2) - \
        #                   (self.chi_e*cos_theta_e-self.chi_c)**2*self.Delta_cos_theta_c_n(cos_theta_c, 1)))

    def Y_s2_l2_m0(self):
        str_Y_s2_l2_m0 = '(3/4)*sqrt(5/(6*pi))*sin(theta_e)**2'
        return sympify(str_Y_s2_l2_m0, evaluate=False)
```
